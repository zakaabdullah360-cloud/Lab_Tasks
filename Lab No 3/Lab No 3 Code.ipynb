```python
!pip -q install kagglehub opencv-python scikit-image seaborn

import os
import time
import random
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import cv2

from PIL import Image

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
from sklearn.decomposition import PCA

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms

import kagglehub

# ============================================================
# 2. REPRODUCIBILITY
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("=" * 70)
print("DEVICE:", device)
print("=" * 70)


# ============================================================
# 3. SETTINGS
# ============================================================

IMAGE_SIZE = 224

# Four classes will automatically be selected based on
# the largest number of images.
NUMBER_OF_CLASSES = 4

# Train / Validation / Test
TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
TEST_RATIO = 0.15

# CNN training
EPOCHS = 5
BATCH_SIZE = 32
LEARNING_RATE = 1e-4

BEST_FILTER_FROM_LAB2 = "Gaussian"

# Best edge detector for classification.
# Canny is used as the default because Lab 03 specifically
# analyzes Canny configurations.
BEST_EDGE_METHOD = "Canny"

# Canny configuration selected for classification
CANNY_LOW = 50
CANNY_HIGH = 150
CANNY_KERNEL = 3

print("\nSettings:")
print("Image Size:", IMAGE_SIZE)
print("Classes:", NUMBER_OF_CLASSES)
print("Best Filter from Lab 02:", BEST_FILTER_FROM_LAB2)
print("Best Edge Method:", BEST_EDGE_METHOD)
print("Canny:", CANNY_LOW, CANNY_HIGH, "Kernel =", CANNY_KERNEL)


# ============================================================
# 4. DOWNLOAD DATASET
# ============================================================

print("\n" + "=" * 70)
print("DOWNLOADING KAGGLE DATASET")
print("=" * 70)

path = kagglehub.dataset_download(
    "nodoubttome/skin-cancer9-classesisic"
)

print("Dataset downloaded to:")
print(path)


# ============================================================
# 5. FIND IMAGE FILES AND CLASSES
# ============================================================

VALID_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
)

class_to_files = {}

for root, dirs, files in os.walk(path):

    image_files = [
        os.path.join(root, f)
        for f in files
        if f.lower().endswith(VALID_EXTENSIONS)
    ]

    if len(image_files) > 0:

        class_name = os.path.basename(root)

        if class_name not in class_to_files:
            class_to_files[class_name] = []

        class_to_files[class_name].extend(image_files)


# Remove empty classes
class_to_files = {
    k: v for k, v in class_to_files.items()
    if len(v) > 0
}

print("\nAll detected classes:")
for cls, files in class_to_files.items():
    print(f"{cls:30s} : {len(files)} images")


# ============================================================
# 6. SELECT FOUR LARGEST CLASSES
# ============================================================

sorted_classes = sorted(
    class_to_files.items(),
    key=lambda x: len(x[1]),
    reverse=True
)

selected_classes = [
    cls for cls, files in sorted_classes[:NUMBER_OF_CLASSES]
]

print("\n" + "=" * 70)
print("SELECTED CLASSES")
print("=" * 70)

for i, cls in enumerate(selected_classes):
    print(
        f"{i}: {cls} -> "
        f"{len(class_to_files[cls])} images"
    )


# ============================================================
# 7. CREATE DATAFRAME
# ============================================================

records = []

for label, class_name in enumerate(selected_classes):

    for file_path in class_to_files[class_name]:

        records.append({
            "filepath": file_path,
            "class": class_name,
            "label": label
        })

df = pd.DataFrame(records)

print("\nTotal selected images:", len(df))

print("\nClass Distribution:")
print(df["class"].value_counts())


# ============================================================
# 8. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

train_df, temp_df = train_test_split(
    df,
    test_size=(VAL_RATIO + TEST_RATIO),
    stratify=df["label"],
    random_state=SEED
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=TEST_RATIO / (VAL_RATIO + TEST_RATIO),
    stratify=temp_df["label"],
    random_state=SEED
)

train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

print("\n" + "=" * 70)
print("DATASET SPLIT")
print("=" * 70)

print("Training:", len(train_df))
print("Validation:", len(val_df))
print("Testing:", len(test_df))


# ============================================================
# 9. DISPLAY CLASS DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="class",
    order=selected_classes
)

plt.title("Class Distribution")
plt.xlabel("Class")
plt.ylabel("Number of Images")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ============================================================
# 10. IMAGE LOADING FUNCTION
# ============================================================

def load_image(path, size=IMAGE_SIZE):

    image = cv2.imread(path)

    if image is None:
        raise ValueError(f"Could not read image: {path}")

    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    image = cv2.resize(
        image,
        (size, size)
    )

    return image


# ============================================================
# 11. EDGE DETECTION FUNCTIONS
# ============================================================

def sobel_gx(image):

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_RGB2GRAY
    )

    gx = cv2.Sobel(
        gray,
        cv2.CV_64F,
        1,
        0,
        ksize=3
    )

    gx = cv2.convertScaleAbs(gx)

    return gx


def sobel_gy(image):

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_RGB2GRAY
    )

    gy = cv2.Sobel(
        gray,
        cv2.CV_64F,
        0,
        1,
        ksize=3
    )

    gy = cv2.convertScaleAbs(gy)

    return gy


def sobel_magnitude(image):

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_RGB2GRAY
    )

    gx = cv2.Sobel(
        gray,
        cv2.CV_64F,
        1,
        0,
        ksize=3
    )

    gy = cv2.Sobel(
        gray,
        cv2.CV_64F,
        0,
        1,
        ksize=3
    )

    magnitude = cv2.magnitude(
        gx.astype(np.float32),
        gy.astype(np.float32)
    )

    magnitude = cv2.normalize(
        magnitude,
        None,
        0,
        255,
        cv2.NORM_MINMAX
    )

    return magnitude.astype(np.uint8)


def prewitt(image):

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_RGB2GRAY
    )

    kernel_x = np.array([
        [-1, 0, 1],
        [-1, 0, 1],
        [-1, 0, 1]
    ], dtype=np.float32)

    kernel_y = np.array([
        [-1, -1, -1],
        [0, 0, 0],
        [1, 1, 1]
    ], dtype=np.float32)

    px = cv2.filter2D(
        gray,
        cv2.CV_32F,
        kernel_x
    )

    py = cv2.filter2D(
        gray,
        cv2.CV_32F,
        kernel_y
    )

    magnitude = cv2.magnitude(
        px,
        py
    )

    magnitude = cv2.normalize(
        magnitude,
        None,
        0,
        255,
        cv2.NORM_MINMAX
    )

    return magnitude.astype(np.uint8)


def laplacian(image):

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_RGB2GRAY
    )

    result = cv2.Laplacian(
        gray,
        cv2.CV_64F,
        ksize=3
    )

    result = cv2.convertScaleAbs(result)

    return result


def log_detector(image):

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_RGB2GRAY
    )

    blurred = cv2.GaussianBlur(
        gray,
        (5, 5),
        0
    )

    result = cv2.Laplacian(
        blurred,
        cv2.CV_64F,
        ksize=3
    )

    result = cv2.convertScaleAbs(result)

    return result


def canny(image, low=50, high=150):

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_RGB2GRAY
    )

    result = cv2.Canny(
        gray,
        low,
        high,
        apertureSize=CANNY_KERNEL
    )

    return result


# ============================================================
# 12. TASK 1
# EDGE DETECTION VISUALIZATION
# ============================================================

print("\n" + "=" * 70)
print("TASK 1 - EDGE DETECTION")
print("=" * 70)

# One image from at least three classes
sample_images = []

for cls in selected_classes[:3]:

    row = df[df["class"] == cls].iloc[0]

    sample_images.append(
        (
            cls,
            load_image(row["filepath"])
        )
    )


for class_name, image in sample_images:

    gx = sobel_gx(image)
    gy = sobel_gy(image)
    sm = sobel_magnitude(image)
    pw = prewitt(image)
    lp = laplacian(image)
    lg = log_detector(image)
    cn = canny(
        image,
        CANNY_LOW,
        CANNY_HIGH
    )

    plt.figure(figsize=(18, 8))

    plt.subplot(2, 4, 1)
    plt.imshow(image)
    plt.title(f"Original\n{class_name}")
    plt.axis("off")

    plt.subplot(2, 4, 2)
    plt.imshow(gx, cmap="gray")
    plt.title("Sobel Gx")
    plt.axis("off")

    plt.subplot(2, 4, 3)
    plt.imshow(gy, cmap="gray")
    plt.title("Sobel Gy")
    plt.axis("off")

    plt.subplot(2, 4, 4)
    plt.imshow(sm, cmap="gray")
    plt.title("Sobel Magnitude")
    plt.axis("off")

    plt.subplot(2, 4, 5)
    plt.imshow(pw, cmap="gray")
    plt.title("Prewitt")
    plt.axis("off")

    plt.subplot(2, 4, 6)
    plt.imshow(lp, cmap="gray")
    plt.title("Laplacian")
    plt.axis("off")

    plt.subplot(2, 4, 7)
    plt.imshow(lg, cmap="gray")
    plt.title("LoG")
    plt.axis("off")

    plt.subplot(2, 4, 8)
    plt.imshow(cn, cmap="gray")
    plt.title("Canny")
    plt.axis("off")

    plt.suptitle(
        f"Edge Detection - {class_name}",
        fontsize=16
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 13. NOISE FUNCTIONS
# ============================================================

def add_gaussian_noise(image):

    image_float = image.astype(np.float32)

    noise = np.random.normal(
        0,
        25,
        image.shape
    )

    noisy = image_float + noise

    noisy = np.clip(
        noisy,
        0,
        255
    ).astype(np.uint8)

    return noisy


def add_salt_pepper_noise(
    image,
    amount=0.04,
    salt_vs_pepper=0.5
):

    noisy = image.copy()

    total_pixels = image.shape[0] * image.shape[1]

    num_salt = int(
        total_pixels * amount * salt_vs_pepper
    )

    num_pepper = int(
        total_pixels * amount * (1 - salt_vs_pepper)
    )

    # Salt
    coords = (
        np.random.randint(0, image.shape[0], num_salt),
        np.random.randint(0, image.shape[1], num_salt)
    )

    noisy[coords[0], coords[1]] = 255

    # Pepper
    coords = (
        np.random.randint(0, image.shape[0], num_pepper),
        np.random.randint(0, image.shape[1], num_pepper)
    )

    noisy[coords[0], coords[1]] = 0

    return noisy


# ============================================================
# 14. FILTER FUNCTIONS
# ============================================================

def gaussian_filter(image):

    return cv2.GaussianBlur(
        image,
        (5, 5),
        0
    )


def average_filter(image):

    return cv2.blur(
        image,
        (5, 5)
    )


def median_filter(image):

    return cv2.medianBlur(
        image,
        5
    )


def sharpening_filter(image):

    kernel = np.array([
        [0, -1, 0],
        [-1, 5, -1],
        [0, -1, 0]
    ])

    return cv2.filter2D(
        image,
        -1,
        kernel
    )


# ============================================================
# 15. EDGE QUALITY MEASURE
# ============================================================

def edge_density(edge_image):

    if edge_image is None:
        return 0

    return np.mean(edge_image > 0)


def edge_quality_score(edge_image):

    """
    Simple reproducible score for comparison.

    It rewards:
    - useful edge pixels
    - avoids extremely noisy maps

    This is only a computational comparison.
    Visual inspection is also required for the report.
    """

    density = edge_density(edge_image)

    # Reasonable edge density is approximately 5%-30%
    if density < 0.05:
        score = density / 0.05
    elif density <= 0.30:
        score = 1.0
    else:
        score = max(
            0,
            1 - (density - 0.30)
        )

    return score


# ============================================================
# 16. TASK 2
# NOISE + EDGE DETECTION
# ============================================================

print("\n" + "=" * 70)
print("TASK 2 - NOISE ANALYSIS")
print("=" * 70)

noise_class = selected_classes[0]

noise_sample_path = df[
    df["class"] == noise_class
].iloc[0]["filepath"]

original_noise_image = load_image(
    noise_sample_path
)

gaussian_noisy = add_gaussian_noise(
    original_noise_image
)

salt_pepper_noisy = add_salt_pepper_noise(
    original_noise_image
)

gaussian_filtered = gaussian_filter(
    gaussian_noisy
)

gaussian_median = median_filter(
    gaussian_noisy
)

sp_gaussian_filtered = gaussian_filter(
    salt_pepper_noisy
)

sp_median_filtered = median_filter(
    salt_pepper_noisy
)


# ------------------------------------------------------------
# Visual comparison
# ------------------------------------------------------------

noise_inputs = [
    ("Original", original_noise_image),
    ("Gaussian Noise", gaussian_noisy),
    ("Gaussian Noise + Gaussian Filter", gaussian_filtered),
    ("Gaussian Noise + Median Filter", gaussian_median),
    ("Salt & Pepper Noise", salt_pepper_noisy),
    ("S&P + Gaussian Filter", sp_gaussian_filtered),
    ("S&P + Median Filter", sp_median_filtered)
]


for title, img in noise_inputs:

    edges = canny(
        img,
        CANNY_LOW,
        CANNY_HIGH
    )

    plt.figure(figsize=(12, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(img)
    plt.title(title)
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(edges, cmap="gray")
    plt.title("Canny Edge Map")
    plt.axis("off")

    plt.tight_layout()
    plt.show()


# ============================================================
# 17. CREATE TABLE 1
# ============================================================

table1_rows = []

noise_experiments = [
    ("Original", original_noise_image, "None", "None"),
    ("Gaussian", gaussian_noisy, "Gaussian", "None"),
    ("Gaussian", gaussian_filtered, "Gaussian", "Gaussian"),
    ("Gaussian", gaussian_median, "Gaussian", "Median"),
    ("Salt-Pepper", salt_pepper_noisy, "Salt-Pepper", "None"),
    ("Salt-Pepper", sp_gaussian_filtered, "Salt-Pepper", "Gaussian"),
    ("Salt-Pepper", sp_median_filtered, "Salt-Pepper", "Median")
]


detectors = {
    "Sobel": sobel_magnitude,
    "Prewitt": prewitt,
    "Laplacian": laplacian,
    "LoG": log_detector,
    "Canny": lambda x: canny(
        x,
        CANNY_LOW,
        CANNY_HIGH
    )
}


for input_name, img, noise_type, preprocessing in noise_experiments:

    for detector_name, detector_function in detectors.items():

        edge_img = detector_function(img)

        quality = edge_quality_score(
            edge_img
        )

        density = edge_density(
            edge_img
        )

        if density > 0.35:
            sensitivity = "High"
        elif density > 0.20:
            sensitivity = "Medium"
        else:
            sensitivity = "Low"

        if preprocessing == "Median":
            observation = (
                "Median filtering reduces impulse noise "
                "and produces cleaner edges."
            )

        elif preprocessing == "Gaussian":
            observation = (
                "Gaussian filtering smooths noise but "
                "may slightly blur edges."
            )

        elif noise_type == "Salt-Pepper":
            observation = (
                "Salt-and-pepper noise may create false "
                "or broken edges."
            )

        elif noise_type == "Gaussian":
            observation = (
                "Gaussian noise introduces extra edge "
                "responses."
            )

        else:
            observation = (
                "Original image used as baseline."
            )

        table1_rows.append({
            "Edge Detector": detector_name,
            "Input Image": input_name,
            "Noise Type": noise_type,
            "Preprocessing": preprocessing,
            "Edge Quality": round(quality, 3),
            "Noise Sensitivity": sensitivity,
            "Observations": observation
        })


table1 = pd.DataFrame(table1_rows)

print("\n" + "=" * 70)
print("TABLE 1 - NOISE AND EDGE DETECTION")
print("=" * 70)

display(table1)


# ============================================================
# 18. TASK 3
# CANNY PARAMETER ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("TASK 3 - CANNY PARAMETER ANALYSIS")
print("=" * 70)

canny_configs = [
    (30, 100, 3),
    (50, 150, 3),
    (100, 200, 3),
    (50, 150, 5)
]

canny_results = []

for low, high, kernel in canny_configs:

    gray = cv2.cvtColor(
        original_noise_image,
        cv2.COLOR_RGB2GRAY
    )

    edges = cv2.Canny(
        gray,
        low,
        high,
        apertureSize=kernel
    )

    number_of_edges = int(
        np.sum(edges > 0)
    )

    density = np.mean(
        edges > 0
    )

    quality = edge_quality_score(
        edges
    )

    if quality >= 0.9:
        observation = "Good edge representation"

    elif quality >= 0.6:
        observation = "Moderate edge representation"

    else:
        observation = "Low edge representation"

    canny_results.append({
        "Configuration": f"{low}-{high}",
        "Low Threshold": low,
        "High Threshold": high,
        "Kernel Size": f"{kernel}x{kernel}",
        "Edge Quality": round(quality, 3),
        "Number of Detected Edges": number_of_edges,
        "Observation": observation
    })

    plt.figure(figsize=(6, 5))

    plt.imshow(
        edges,
        cmap="gray"
    )

    plt.title(
        f"Canny: Low={low}, High={high}, "
        f"Kernel={kernel}x{kernel}"
    )

    plt.axis("off")
    plt.show()


table2 = pd.DataFrame(canny_results)

print("\n" + "=" * 70)
print("TABLE 2 - CANNY PARAMETER ANALYSIS")
print("=" * 70)

display(table2)


# ============================================================
# 19. SELECT BEST CANNY CONFIGURATION
# ============================================================

best_canny_row = table2.loc[
    table2["Edge Quality"].idxmax()
]

BEST_CANNY_LOW = int(
    best_canny_row["Low Threshold"]
)

BEST_CANNY_HIGH = int(
    best_canny_row["High Threshold"]
)

BEST_CANNY_KERNEL = int(
    best_canny_row["Kernel Size"].split("x")[0]
)

print("\nSelected Canny configuration:")
print(
    f"Low = {BEST_CANNY_LOW}, "
    f"High = {BEST_CANNY_HIGH}, "
    f"Kernel = {BEST_CANNY_KERNEL}x{BEST_CANNY_KERNEL}"
)


# ============================================================
# 20. FILTERING FUNCTION FOR CLASSIFICATION
# ============================================================

def apply_best_filter(image):

    if BEST_FILTER_FROM_LAB2 == "Gaussian":

        return gaussian_filter(image)

    elif BEST_FILTER_FROM_LAB2 == "Median":

        return median_filter(image)

    elif BEST_FILTER_FROM_LAB2 == "Average":

        return average_filter(image)

    elif BEST_FILTER_FROM_LAB2 == "Sharpening":

        return sharpening_filter(image)

    elif BEST_FILTER_FROM_LAB2 == "Sobel":

        return sobel_magnitude(image)

    else:

        return image


# ============================================================
# 21. EDGE IMAGE FUNCTION FOR CLASSIFICATION
# ============================================================

def create_edge_image(image):

    edges = cv2.Canny(
        cv2.cvtColor(
            image,
            cv2.COLOR_RGB2GRAY
        ),
        BEST_CANNY_LOW,
        BEST_CANNY_HIGH,
        apertureSize=BEST_CANNY_KERNEL
    )

    # Convert grayscale edge image to RGB
    edges_rgb = cv2.cvtColor(
        edges,
        cv2.COLOR_GRAY2RGB
    )

    return edges_rgb


# ============================================================
# 22. PREPARE CLASSIFICATION DATA
# ============================================================

print("\n" + "=" * 70)
print("TASK 4 - PREPARING RAW / FILTERED / EDGE DATA")
print("=" * 70)


def prepare_numpy_dataset(dataframe, representation):

    images = []
    labels = []

    for _, row in dataframe.iterrows():

        image = load_image(
            row["filepath"]
        )

        if representation == "Raw":

            processed = image

        elif representation == "Filtered":

            processed = apply_best_filter(
                image
            )

            # If filter is Sobel, convert to RGB
            if len(processed.shape) == 2:
                processed = cv2.cvtColor(
                    processed,
                    cv2.COLOR_GRAY2RGB
                )

        elif representation == "Edge":

            processed = create_edge_image(
                image
            )

        images.append(
            processed
        )

        labels.append(
            row["label"]
        )

    return (
        np.array(images),
        np.array(labels)
    )


print("\nPreparing RAW data...")

X_train_raw, y_train = prepare_numpy_dataset(
    train_df,
    "Raw"
)

X_val_raw, y_val = prepare_numpy_dataset(
    val_df,
    "Raw"
)

X_test_raw, y_test = prepare_numpy_dataset(
    test_df,
    "Raw"
)


print("Preparing FILTERED data...")

X_train_filtered, _ = prepare_numpy_dataset(
    train_df,
    "Filtered"
)

X_val_filtered, _ = prepare_numpy_dataset(
    val_df,
    "Filtered"
)

X_test_filtered, _ = prepare_numpy_dataset(
    test_df,
    "Filtered"
)


print("Preparing EDGE data...")

X_train_edge, _ = prepare_numpy_dataset(
    train_df,
    "Edge"
)

X_val_edge, _ = prepare_numpy_dataset(
    val_df,
    "Edge"
)

X_test_edge, _ = prepare_numpy_dataset(
    test_df,
    "Edge"
)


print("\nShapes:")

print(
    "Raw:",
    X_train_raw.shape,
    X_val_raw.shape,
    X_test_raw.shape
)

print(
    "Filtered:",
    X_train_filtered.shape,
    X_val_filtered.shape,
    X_test_filtered.shape
)

print(
    "Edge:",
    X_train_edge.shape,
    X_val_edge.shape,
    X_test_edge.shape
)


# ============================================================
# 23. VISUALIZE RAW / FILTERED / EDGE
# ============================================================

sample_idx = 0

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)

plt.imshow(
    X_test_raw[sample_idx]
)

plt.title("Set A - Raw")
plt.axis("off")


plt.subplot(1, 3, 2)

plt.imshow(
    X_test_filtered[sample_idx]
)

plt.title(
    f"Set B - Filtered ({BEST_FILTER_FROM_LAB2})"
)

plt.axis("off")


plt.subplot(1, 3, 3)

plt.imshow(
    X_test_edge[sample_idx],
    cmap="gray"
)

plt.title(
    f"Set C - Edge ({BEST_EDGE_METHOD})"
)

plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 24. PREPARE HOG FEATURES FOR CLASSICAL ML
# ============================================================

from skimage.feature import hog


def extract_hog_features(images):

    features = []

    for image in images:

        gray = cv2.cvtColor(
            image,
            cv2.COLOR_RGB2GRAY
        )

        feature = hog(
            gray,
            orientations=9,
            pixels_per_cell=(16, 16),
            cells_per_block=(2, 2),
            block_norm="L2-Hys"
        )

        features.append(feature)

    return np.array(features)


print("\n" + "=" * 70)
print("EXTRACTING HOG FEATURES")
print("=" * 70)


X_train_raw_hog = extract_hog_features(
    X_train_raw
)

X_test_raw_hog = extract_hog_features(
    X_test_raw
)

X_train_filtered_hog = extract_hog_features(
    X_train_filtered
)

X_test_filtered_hog = extract_hog_features(
    X_test_filtered
)

X_train_edge_hog = extract_hog_features(
    X_train_edge
)

X_test_edge_hog = extract_hog_features(
    X_test_edge
)


print(
    "HOG feature size:",
    X_train_raw_hog.shape
)


# ============================================================
# 25. CLASSICAL MACHINE LEARNING MODELS
# ============================================================

def evaluate_classical_model(
    model,
    X_train,
    y_train,
    X_test,
    y_test
):

    start_train = time.time()

    model.fit(
        X_train,
        y_train
    )

    training_time = (
        time.time() - start_train
    )

    start_inference = time.time()

    predictions = model.predict(
        X_test
    )

    inference_time = (
        time.time() - start_inference
    )

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    precision = precision_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    return {
        "model": model,
        "predictions": predictions,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "training_time": training_time,
        "inference_time_ms": (
            inference_time / len(y_test)
        ) * 1000
    }


def create_models():

    return {
        "SVM": Pipeline([
            (
                "scaler",
                StandardScaler()
            ),
            (
                "classifier",
                SVC(
                    kernel="rbf",
                    C=10,
                    gamma="scale"
                )
            )
        ]),

        "Random Forest": RandomForestClassifier(
            n_estimators=100,
            random_state=SEED,
            n_jobs=-1
        ),

        "KNN": Pipeline([
            (
                "scaler",
                StandardScaler()
            ),
            (
                "classifier",
                KNeighborsClassifier(
                    n_neighbors=5
                )
            )
        ])
    }


# ============================================================
# 26. RUN CLASSICAL MODELS
# ============================================================

classical_results = []

classical_predictions = {}

representations = {
    "Raw": (
        X_train_raw_hog,
        X_test_raw_hog
    ),

    "Filtered": (
        X_train_filtered_hog,
        X_test_filtered_hog
    ),

    "Edge": (
        X_train_edge_hog,
        X_test_edge_hog
    )
}


for representation_name, (
    Xtr,
    Xte
) in representations.items():

    print("\n" + "-" * 60)
    print("Representation:", representation_name)
    print("-" * 60)

    for model_name, model in create_models().items():

        print(
            "Training",
            model_name,
            "..."
        )

        result = evaluate_classical_model(
            model,
            Xtr,
            y_train,
            Xte,
            y_test
        )

        classical_results.append({
            "Model": model_name,
            "Representation": representation_name,
            "Accuracy": result["accuracy"],
            "Precision": result["precision"],
            "Recall": result["recall"],
            "F1": result["f1"],
            "Training Time(s)": result["training_time"],
            "Inference Time(ms)": result["inference_time_ms"]
        })

        classical_predictions[
            (model_name, representation_name)
        ] = result["predictions"]

        print(
            f"Accuracy: {result['accuracy']:.4f}"
        )


classical_df = pd.DataFrame(
    classical_results
)

print("\n" + "=" * 70)
print("CLASSICAL MODEL RESULTS")
print("=" * 70)

display(
    classical_df.round(4)
)


# ============================================================
# 27. CNN DATASET CLASS
# ============================================================

class ImageArrayDataset(Dataset):

    def __init__(
        self,
        images,
        labels,
        train=False
    ):

        self.images = images
        self.labels = labels
        self.train = train

        if train:

            self.transform = transforms.Compose([
                transforms.ToPILImage(),

                transforms.RandomHorizontalFlip(),

                transforms.RandomRotation(10),

                transforms.ToTensor(),

                transforms.Normalize(
                    mean=[
                        0.485,
                        0.456,
                        0.406
                    ],
                    std=[
                        0.229,
                        0.224,
                        0.225
                    ]
                )
            ])

        else:

            self.transform = transforms.Compose([
                transforms.ToPILImage(),

                transforms.ToTensor(),

                transforms.Normalize(
                    mean=[
                        0.485,
                        0.456,
                        0.406
                    ],
                    std=[
                        0.229,
                        0.224,
                        0.225
                    ]
                )
            ])

    def __len__(self):

        return len(self.images)

    def __getitem__(self, idx):

        image = self.images[idx]
        label = self.labels[idx]

        image = self.transform(
            image
        )

        return image, torch.tensor(
            label,
            dtype=torch.long
        )


# ============================================================
# 28. CNN TRAINING FUNCTION
# ============================================================

def train_cnn(
    model,
    train_loader,
    val_loader,
    epochs=EPOCHS
):

    criterion = nn.CrossEntropyLoss()

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=LEARNING_RATE
    )

    model.to(device)

    history = {
        "train_loss": [],
        "val_loss": [],
        "train_acc": [],
        "val_acc": []
    }

    start_time = time.time()

    for epoch in range(epochs):

        model.train()

        running_loss = 0
        correct = 0
        total = 0

        for images, labels in train_loader:

            images = images.to(device)
            labels = labels.to(device)

            optimizer.zero_grad()

            outputs = model(
                images
            )

            loss = criterion(
                outputs,
                labels
            )

            loss.backward()

            optimizer.step()

            running_loss += (
                loss.item()
                * images.size(0)
            )

            _, predicted = torch.max(
                outputs,
                1
            )

            total += labels.size(0)

            correct += (
                predicted == labels
            ).sum().item()

        train_loss = (
            running_loss / total
        )

        train_acc = (
            correct / total
        )

        # Validation
        model.eval()

        val_loss_total = 0
        val_correct = 0
        val_total = 0

        with torch.no_grad():

            for images, labels in val_loader:

                images = images.to(device)
                labels = labels.to(device)

                outputs = model(
                    images
                )

                loss = criterion(
                    outputs,
                    labels
                )

                val_loss_total += (
                    loss.item()
                    * images.size(0)
                )

                _, predicted = torch.max(
                    outputs,
                    1
                )

                val_total += labels.size(0)

                val_correct += (
                    predicted == labels
                ).sum().item()

        val_loss = (
            val_loss_total / val_total
        )

        val_acc = (
            val_correct / val_total
        )

        history["train_loss"].append(
            train_loss
        )

        history["val_loss"].append(
            val_loss
        )

        history["train_acc"].append(
            train_acc
        )

        history["val_acc"].append(
            val_acc
        )

        print(
            f"Epoch {epoch+1}/{epochs} | "
            f"Train Acc: {train_acc:.4f} | "
            f"Val Acc: {val_acc:.4f}"
        )

    training_time = (
        time.time() - start_time
    )

    return (
        model,
        history,
        training_time
    )


# ============================================================
# 29. CNN EVALUATION FUNCTION
# ============================================================

def evaluate_cnn(
    model,
    test_loader,
    y_true
):

    model.eval()

    predictions = []

    start_time = time.time()

    with torch.no_grad():

        for images, labels in test_loader:

            images = images.to(device)

            outputs = model(
                images
            )

            _, predicted = torch.max(
                outputs,
                1
            )

            predictions.extend(
                predicted.cpu().numpy()
            )

    inference_time = (
        time.time() - start_time
    )

    predictions = np.array(
        predictions
    )

    accuracy = accuracy_score(
        y_true,
        predictions
    )

    precision = precision_score(
        y_true,
        predictions,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_true,
        predictions,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        predictions,
        average="weighted",
        zero_division=0
    )

    return {
        "predictions": predictions,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "inference_time_ms": (
            inference_time / len(y_true)
        ) * 1000
    }


# ============================================================
# 30. CNN MODEL CREATION
# ============================================================

def create_resnet18():

    model = models.resnet18(
        weights=models.ResNet18_Weights.DEFAULT
    )

    model.fc = nn.Linear(
        model.fc.in_features,
        NUMBER_OF_CLASSES
    )

    return model


def create_mobilenet():

    model = models.mobilenet_v2(
        weights=models.MobileNet_V2_Weights.DEFAULT
    )

    model.classifier[1] = nn.Linear(
        model.classifier[1].in_features,
        NUMBER_OF_CLASSES
    )

    return model


# ============================================================
# 31. CNN TRAINING
# ============================================================

cnn_results = []

cnn_predictions = {}

cnn_histories = {}

cnn_datasets = {

    "Raw": (
        X_train_raw,
        X_test_raw
    ),

    "Filtered": (
        X_train_filtered,
        X_test_filtered
    ),

    "Edge": (
        X_train_edge,
        X_test_edge
    )
}


# To reduce excessive training time,
# use two CNN models as required by Lab 03.

cnn_model_builders = {
    "CNN Model 1 - ResNet18": create_resnet18,
    "CNN Model 2 - MobileNetV2": create_mobilenet
}


for model_name, model_builder in cnn_model_builders.items():

    for representation_name, (
        Xtr,
        Xte
    ) in cnn_datasets.items():

        print("\n")
        print("=" * 70)
        print(
            model_name,
            "|",
            representation_name
        )
        print("=" * 70)

        train_dataset = ImageArrayDataset(
            Xtr,
            y_train,
            train=True
        )

        val_dataset = ImageArrayDataset(
            Xtr[:max(1, int(len(Xtr) * 0.15))],
            y_train[:max(1, int(len(y_train) * 0.15))],
            train=False
        )

        test_dataset = ImageArrayDataset(
            Xte,
            y_test,
            train=False
        )

        train_loader = DataLoader(
            train_dataset,
            batch_size=BATCH_SIZE,
            shuffle=True,
            num_workers=2,
            pin_memory=True
        )

        val_loader = DataLoader(
            val_dataset,
            batch_size=BATCH_SIZE,
            shuffle=False,
            num_workers=2,
            pin_memory=True
        )

        test_loader = DataLoader(
            test_dataset,
            batch_size=BATCH_SIZE,
            shuffle=False,
            num_workers=2,
            pin_memory=True
        )

        model = model_builder()

        model, history, training_time = train_cnn(
            model,
            train_loader,
            val_loader,
            EPOCHS
        )

        result = evaluate_cnn(
            model,
            test_loader,
            y_test
        )

        cnn_results.append({
            "Model": model_name,
            "Representation": representation_name,
            "Accuracy": result["accuracy"],
            "Precision": result["precision"],
            "Recall": result["recall"],
            "F1": result["f1"],
            "Training Time(s)": training_time,
            "Inference Time(ms)": result["inference_time_ms"]
        })

        cnn_predictions[
            (model_name, representation_name)
        ] = result["predictions"]

        cnn_histories[
            (model_name, representation_name)
        ] = history

        print(
            f"\nTest Accuracy: "
            f"{result['accuracy']:.4f}"
        )


cnn_df = pd.DataFrame(
    cnn_results
)

print("\n" + "=" * 70)
print("CNN RESULTS")
print("=" * 70)

display(
    cnn_df.round(4)
)


# ============================================================
# 32. COMBINE ALL RESULTS
# ============================================================

all_results = pd.concat(
    [
        classical_df,
        cnn_df
    ],
    ignore_index=True
)

print("\n" + "=" * 70)
print("ALL CLASSIFICATION RESULTS")
print("=" * 70)

display(
    all_results.round(4)
)


# ============================================================
# 33. CREATE TABLE 3
# ============================================================

table3_rows = []

all_model_names = all_results[
    "Model"
].unique()


for model_name in all_model_names:

    model_data = all_results[
        all_results["Model"] == model_name
    ]

    raw = model_data[
        model_data["Representation"] == "Raw"
    ]

    filtered = model_data[
        model_data["Representation"] == "Filtered"
    ]

    edge = model_data[
        model_data["Representation"] == "Edge"
    ]

    if len(raw) == 0 or len(filtered) == 0 or len(edge) == 0:
        continue

    # Use EDGE representation for the detailed
    # Precision / Recall / F1 fields in the
    # requested compact Table 3.

    edge_row = edge.iloc[0]

    table3_rows.append({
        "Model / Classifier": model_name,

        "Accuracy Raw":
            raw.iloc[0]["Accuracy"],

        "Accuracy Filtered":
            filtered.iloc[0]["Accuracy"],

        "Accuracy Edge":
            edge.iloc[0]["Accuracy"],

        "Precision":
            edge_row["Precision"],

        "Recall":
            edge_row["Recall"],

        "F1":
            edge_row["F1"],

        "Training Time(s)":
            edge_row["Training Time(s)"],

        "Inference Time(ms)":
            edge_row["Inference Time(ms)"]
    })


table3 = pd.DataFrame(
    table3_rows
)

print("\n" + "=" * 70)
print("TABLE 3 - RAW VS FILTERED VS EDGE CLASSIFICATION")
print("=" * 70)

display(
    table3.round(4)
)


# ============================================================
# 34. BEST MODEL
# ============================================================

best_row = all_results.loc[
    all_results["Accuracy"].idxmax()
]

print("\n" + "=" * 70)
print("BEST PERFORMING MODEL / REPRESENTATION")
print("=" * 70)

print(
    "Model:",
    best_row["Model"]
)

print(
    "Representation:",
    best_row["Representation"]
)

print(
    "Accuracy:",
    round(best_row["Accuracy"], 4)
)

print(
    "Precision:",
    round(best_row["Precision"], 4)
)

print(
    "Recall:",
    round(best_row["Recall"], 4)
)

print(
    "F1:",
    round(best_row["F1"], 4)
)


# ============================================================
# 35. CONFUSION MATRICES
# ============================================================

print("\n" + "=" * 70)
print("TASK 6 - CONFUSION MATRICES")
print("=" * 70)


best_model_name = best_row["Model"]

best_model_data = all_results[
    all_results["Model"] == best_model_name
]

for representation in [
    "Raw",
    "Filtered",
    "Edge"
]:

    row = best_model_data[
        best_model_data["Representation"]
        == representation
    ]

    if len(row) == 0:
        continue

    if best_model_name in [
        "SVM",
        "Random Forest",
        "KNN"
    ]:

        predictions = classical_predictions[
            (
                best_model_name,
                representation
            )
        ]

    else:

        predictions = cnn_predictions[
            (
                best_model_name,
                representation
            )
        ]

    cm = confusion_matrix(
        y_test,
        predictions
    )

    plt.figure(figsize=(7, 6))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=selected_classes,
        yticklabels=selected_classes
    )

    plt.title(
        f"Confusion Matrix\n"
        f"{best_model_name} - {representation}"
    )

    plt.xlabel("Predicted")
    plt.ylabel("Actual")

    plt.xticks(rotation=30)
    plt.yticks(rotation=0)

    plt.tight_layout()
    plt.show()


# ============================================================
# 36. ACCURACY COMPARISON
# ============================================================

accuracy_pivot = all_results.pivot(
    index="Model",
    columns="Representation",
    values="Accuracy"
)

plt.figure(figsize=(12, 6))

accuracy_pivot.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title(
    "Accuracy Comparison: Raw vs Filtered vs Edge"
)

plt.xlabel("Model / Classifier")
plt.ylabel("Accuracy")

plt.xticks(rotation=30)
plt.ylim(0, 1)

plt.legend(
    title="Representation"
)

plt.tight_layout()
plt.show()


# ============================================================
# 37. PRECISION / RECALL / F1 COMPARISON
# ============================================================

best_representation_results = all_results[
    all_results["Representation"] == "Edge"
].copy()

metrics_df = best_representation_results[
    [
        "Model",
        "Precision",
        "Recall",
        "F1"
    ]
].set_index("Model")


metrics_df.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title(
    "Precision, Recall and F1 - Edge Representation"
)

plt.xlabel("Model / Classifier")
plt.ylabel("Score")

plt.ylim(0, 1)

plt.xticks(rotation=30)

plt.legend(
    title="Metric"
)

plt.tight_layout()
plt.show()


# ============================================================
# 38. CNN TRAINING CURVES
# ============================================================

for key, history in cnn_histories.items():

    model_name, representation = key

    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)

    plt.plot(
        history["train_loss"],
        label="Training Loss"
    )

    plt.plot(
        history["val_loss"],
        label="Validation Loss"
    )

    plt.title(
        f"Loss - {model_name}\n{representation}"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Loss")

    plt.legend()


    plt.subplot(1, 2, 2)

    plt.plot(
        history["train_acc"],
        label="Training Accuracy"
    )

    plt.plot(
        history["val_acc"],
        label="Validation Accuracy"
    )

    plt.title(
        f"Accuracy - {model_name}\n{representation}"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")

    plt.legend()

    plt.tight_layout()
    plt.show()


# ============================================================
# 39. SAVE TABLES TO CSV
# ============================================================

table1.to_csv(
    "Lab03_Table1_Noise_Edge.csv",
    index=False
)

table2.to_csv(
    "Lab03_Table2_Canny.csv",
    index=False
)

table3.to_csv(
    "Lab03_Table3_Classification.csv",
    index=False
)

all_results.to_csv(
    "Lab03_All_Classification_Results.csv",
    index=False
)


# ============================================================
# 40. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("LAB 03 COMPLETED")
print("=" * 70)

print("\nSelected Classes:")

for cls in selected_classes:
    print("-", cls)

print("\nDataset Split:")
print(
    f"Train: {len(train_df)} "
    f"({len(train_df)/len(df)*100:.1f}%)"
)

print(
    f"Validation: {len(val_df)} "
    f"({len(val_df)/len(df)*100:.1f}%)"
)

print(
    f"Test: {len(test_df)} "
    f"({len(test_df)/len(df)*100:.1f}%)"
)

print("\nBest Filter from Lab 02:")
print(BEST_FILTER_FROM_LAB2)

print("\nSelected Canny Configuration:")
print(
    f"Low={BEST_CANNY_LOW}, "
    f"High={BEST_CANNY_HIGH}, "
    f"Kernel={BEST_CANNY_KERNEL}x{BEST_CANNY_KERNEL}"
)

print("\nBest Overall Result:")
print(
    f"{best_row['Model']} "
    f"({best_row['Representation']})"
)

print(
    f"Accuracy = {best_row['Accuracy']:.4f}"
)

print("\nGenerated Files:")
print("- Lab03_Table1_Noise_Edge.csv")
print("- Lab03_Table2_Canny.csv")
print("- Lab03_Table3_Classification.csv")
print("- Lab03_All_Classification_Results.csv")

print("\n" + "=" * 70)
print("END OF LAB 03")
print("=" * 70)
```
